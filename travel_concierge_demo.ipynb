# Travel Itinerary Planning Concierge

Offline Semantic Kernel manager-worker demo. All prices are illustrative estimates.

```python
from travel_concierge.planner_manager import PlannerManager
from travel_concierge.state import AppendOnlyRunLog

request = {
    'origin': 'Boston',
    'destination': 'Kyoto',
    'days': 3,
    'travelers': 2,
    'budget': 900,
    'interests': ['culture', 'food', 'nature'],
}
result = PlannerManager(AppendOnlyRunLog('run_log.jsonl')).build_itinerary(request)
print('ITINERARY')
for day in result['days']:
    print(f"\nDay {day['day']}")
    for item in day['items']:
        print(f"  - {item['schedule'] if 'schedule' in item else 'Activity'}: {item['name']} (${item['cost']:.2f})")
print(f"\nBudget: ${result['budget']['spent']:.2f} / ${result['budget']['budget']:.2f}; remaining ${result['budget']['remaining']:.2f}")
print(f"Re-plans: {result['replans']}")
print('\nMULTI-AGENT TRACE')
for event in result['trace']:
    print(f"{event['agent']}: {event['action']} | { {k: v for k, v in event.items() if k not in ('timestamp', 'agent', 'action')} }")
```
